```python
import yaml
import random
import math

from datasets import load_dataset, load_from_disk
from datasets import Dataset
from tokenizers import Tokenizer, decoders

# standardize relative filepaths
%cd ../../
```

```python
# ========== Config ==========

config_path = "SL/configs/transformer_tinystories_small_v2.yaml"
device = "cuda"
torch_compile_type = "default"

# ========== Init ==========

# open config
with open(config_path) as ConfigFile:
    c = yaml.safe_load(ConfigFile)

# load tokenizer
tokenizer = Tokenizer.from_file(c["tokenizer_path"])
tokenizer.decoder = decoders.ByteLevel()
c["vocab_size"] = tokenizer.get_vocab_size()

# load dataset
dataset = load_from_disk(c["dataset_path"], keep_in_memory=False)
```

```python
print(f"This dataset has {len(dataset) * (c["seq_len"] + 1)} tokens in {len(dataset)} rows using the tokenizer used for preprocessing!")
print(f"The input sequence length is {c["seq_len"]} so it has dimension ({len(dataset)}, {(c["seq_len"] + 1)})!")
print(f"At effective batch size {c["effective_batch_size"]} this results in {len(dataset)}/{c["effective_batch_size"]} = {len(dataset)/c["effective_batch_size"]} steps!")
print(f"For now put max_steps at {math.floor(len(dataset)/c["effective_batch_size"]/10)*10}")
print("Here is a sample:")
for i in range(1):
    sample = dataset[random.randint(0, len(dataset)-1)]
    print(sample)
    print(tokenizer.decode(sample["text"], skip_special_tokens=False))
```
